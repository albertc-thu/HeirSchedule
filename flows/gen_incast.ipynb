```python
import os
import random

incast_degeres = [20, 40, 60, 80, 100, 120]
# incast_degeres = [30, 55, 80, 105, 130, 155, 180, 205, 230, 255]
# incast_degeres = [100]
# incast_degere = 60
flow_size = 100000-1

network_capacity = 0.01
flow_duration = 0.1
host_num = 125

directory_name = f'./flow_data_{host_num}_pureincast/'
os.makedirs(directory_name, exist_ok=True)


for incast_degree in incast_degeres:
    flow_id = 0
    incast_repeat_time = int(100*10**9/8*host_num * flow_duration * network_capacity / flow_size / incast_degree)
    incast_repeat_time = 1 # pure incast
    random_time = [0.0] * incast_repeat_time
    for i in range(incast_repeat_time):
        random_time[i] = random.random() * flow_duration
    print(len(random_time))
    with open(directory_name + "flows_incast_"+str(incast_degree)+ ".txt", 'w') as f:
        for i in range(incast_repeat_time):
            dst = random.randint(0, host_num - 1)
            flags = [0]*host_num
            incast_time = 1.0 + random_time[i]
            for i in range(incast_degree):
                src = random.randint(0, host_num - 1)
                while flags[src] == 1 or dst == src:
                    src = random.randint(0, host_num - 1)
                flags[src] = 1
                flow = str(flow_id) + " " + str(1.000001) + " 0 "+ str(flow_size) + " 0 0 " + str(src) + ' ' + str(dst) + '\n'
                flow_id = flow_id + 1
                f.write(flow)
```
