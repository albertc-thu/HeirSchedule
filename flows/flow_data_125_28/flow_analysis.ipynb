```python
import numpy as np
import matplotlib.pyplot as plt

loads = [0.1, 0.25, 0.5, 0.75, 1]

file_name = "flows_W5_{load}.txt"
for load in loads:
    flow_file_name = file_name.format(load=load)
    flows = np.loadtxt(flow_file_name)
    flows_times = flows[:,1] - 1
    flow_start_time = 0.0
    flow_finish_time = max(flows_times)
    
    epoch_length = 0.00000366 #一个epoch的长度为3.66us
    bin_edges = np.arange(0, flow_finish_time + epoch_length, epoch_length) 
    hist, bins = np.histogram(flows_times, bins=bin_edges)
    print("hist:", hist)
    print("hist_mean: ", np.mean(hist))
    print("hist_sum: ", np.sum(hist))
    # print("bins:", bins)
    # plt.hist(flows_times, bin_edges)
    
```

```python
# locality analysis
# 分析流量分布：Rack内、Pod内、Pod间

import numpy as np
import matplotlib.pyplot as plt
print("hello")
k = 10
k_2 = k/2
k_k_4 = k*k/4

filename = "flows_W5_1.txt"
data = np.loadtxt(filename)
src_dst_pairs = data[:, 6:8]
src_dst_pairs = src_dst_pairs.astype(int)

flowsizes = data[:, 3]
flow_lt_10K  = flowsizes < 10000
flow_lt_100K = flowsizes < 100000
flow_gt_1M = flowsizes > 1000000
flow_lt_512 = flowsizes < 512
# print("flow_lt_10K: ", sum(flow_lt_10K))
# print("flow_lt_512: ", sum(flow_lt_512))
print("flow_lt_10K: ", sum(flow_lt_10K))
print("flow_lt_100K: ", sum(flow_lt_100K))
print("flow_gt_1M: ", sum(flow_gt_1M))
# print("src_dst_pairs: ", src_dst_pairs)
intra_rack = 0
intra_pod = 0
inter_pod = 0

for src, dst in src_dst_pairs:
    # print(src/k_2, dst/k_2, src/k_k_4, dst/k_k_4)
    if src // k_k_4 == dst // k_k_4:
        intra_pod += 1
    else:
        inter_pod += 1

print("intra_rack: ", intra_rack)
print("intra_pod: ", intra_pod)
print("inter_pod: ", inter_pod)
```
