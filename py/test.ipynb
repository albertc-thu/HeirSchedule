```python
import matplotlib.pyplot as plt

x = [2, 4, 6, 8, 10, 12, 15, 16]
y = [1, 4, 10, 19, 32, 47, 74, 85]

plt.plot(x, y, 'ro')
# plt.axis([0, 20, 0, 100])
plt.show()

# 二次函数拟合
import numpy as np
p = np.poly1d(np.polyfit(x, y, 2))
print(p)
plt.plot(x, y, 'ro')
plt.plot(x, p(x), 'g-')
plt.show()
```
